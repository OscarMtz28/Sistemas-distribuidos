## Servidor web

# importamos la libreria

```python
import socket
```

# Definimos el socket 

```python
server_socket = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
```

# Ligamos el puerto al servidor y levantamos el socket

```python
server_socket.bind(('localhost', 8080))
server_socket.listen(1)
print("Serrvidor ejecutandose ....")
```

# Le decimos al socket que podemos aceptar la conexion

```python
conn, addr = server_socket.accept() 
print("Esperando conexion en el puerto {addr}")
```

# Asignamos la informacion por medio del buffer a la variable "data", 

## data.decode convierte la informacion de binario a base 64

```python
data = conn.recv(1024)
print("Mensaje recibido: ", data.decode())
```

# Enviamos la respuesta del servidor

```python
conn.sendall("Hola desde el servidor!!")
conn.close()
```
